```python
import random
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
```

Part A: Reconstructing Problem 3

```python
G = nx.Graph()
G.add_edges_from([("1", "2"), ("2", "3"), ("2", "4"), ("4", "4"), ("4","5"), ("5","1")])
edges = list(G.edges())
print("Number of edges:", len(edges))
print("LOE", edges)
print("Nodes:", list(G.nodes()))

nx.draw(G, with_labels=True, node_size=700, font_color='white', font_weight='bold')
plt.title("Graph A")
```

```python
adjacency_a = {u: sorted(list(G.neighbors(u))) for u in G.nodes()}

for node in sorted(adjacency_a):
    print(f"{node}: {adjacency_a[node]}")
```

Part B: Problem 4 Verification

```python
adjacency_B= {
    1: [2, 5],
    2: [1, 3, 4],
    3: [2],
    4: [2,4,5],
    5: [1,4]
}
Graph_B = nx.Graph()
for node, neighbors in adjacency_B.items():
    for neighbor in neighbors:
        Graph_B.add_edge(node, neighbor)

plt.figure(figsize=(6, 5))
nx.draw(Graph_B, with_labels=True, node_size=700, font_color='white', font_weight='bold')
plt.title("Graph B")
plt.show()
```

```python
nodes_sorted = [1, 2, 3, 4, 5]
adj_matrix_B = nx.to_numpy_array(Graph_B, nodelist=nodes_sorted, dtype=int)
for i in range(len(adj_matrix_B)):
    if adj_matrix_B[i, i] == 1:  # if self-loop exists
        adj_matrix_B[i, i] = 2
    else:
        adj_matrix_B[i, i] = 0

print("Adjacency matrix of B:\n", adj_matrix_B)
```

```python
plt.figure(figsize=(5,5))
plt.spy(adj_matrix_B, markersize=5)
plt.title("Sparsity pattern of adjacency matrix B")
plt.xlabel("Node index")
plt.xticks(ticks=np.arange(len(nodes_sorted)), labels=nodes_sorted)
plt.yticks(ticks=np.arange(len(nodes_sorted)), labels=nodes_sorted)
plt.ylabel("Node index")
plt.show()
```
